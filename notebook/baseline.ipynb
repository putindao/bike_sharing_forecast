```python
import numpy as np
import pandas as pd
from sklearn.metrics import mean_absolute_error
```

```python
daydf = pd.read_csv("../data/raw/day.csv")
```

```python
daydf["baseline_persistence"] = daydf["cnt"].shift(1)
daydf["baseline_seasonal"] = daydf["cnt"].shift(7)

split = int(len(daydf) * 0.8) # train 80%, test 20%
train = daydf.iloc[:split]
test = daydf.iloc[split:]
```

```python
# baseline 1: y chang ngày hôm qua
mae1 = mean_absolute_error(test["cnt"], test["baseline_persistence"])
print(mae1)
```

```python
# baseline 2: giống ngày này tuần trước
mae2 = mean_absolute_error(test['cnt'], test['baseline_seasonal'])
print(mae2)
```

```python
rmse1 = np.sqrt(mae1)
rmse2 = np.sqrt(mae2)
print(rmse1)
print(rmse2)
```
