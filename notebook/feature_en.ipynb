```python
import pandas as pd
```

```python
daydf = pd.read_csv("../data/raw/day.csv")
```

```python
daydf['cnt_lag1'] = daydf['cnt'].shift(1)
daydf['cnt_lag7'] = daydf['cnt'].shift(7)
daydf['cnt_roll7_mean'] = daydf['cnt'].shift(1).rolling(7).mean()
```

```python
daydf = daydf.dropna()
daydf.to_csv("../data/processed/processed_day.csv")
daydf.info()
daydf.head()
```
